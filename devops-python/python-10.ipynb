```python
# OOPS Concepts:

OOPS:
    Object orient programming Structure
    class and object

    class: House - blueprint(4 floors)

    object or instance: 
        house1-  4 floors, 3 bedrooms, 2 bathrooms
        house1-  4 floors, 2 bedrooms, 1 bathrooms
        house1-  4 floors, 4 bedrooms, 3 bathrooms

function - it will depend on instance
method - it depends on the instance
```

```python
class Server:
    server_count = 0 # class attribute
    def __init__(self,name,ip,cpu_usage):
        self.server_name = name   #instance attribute    
        self.server_ip = ip
        self.server_cpu_usage = cpu_usage
        Server.server_count += 1
    
    def get_server_details(self):
        return f"Server Name: {self.server_name}, IP : {self.server_ip}, CPU_Usage: {self.server_cpu_usage}"

web_server= Server('webserver','10.1.2.12',65)

print(web_server.get_server_details())
print(web_server.server_count)

db_server= Server('dbserver','10.4.2.13',45)

print(db_server.get_server_details())
print(db_server.server_count)
```
