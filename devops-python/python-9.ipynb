```python
'''
file handling:

read()- entire lines of code string format
read(size)- number of characters
readline()- will return the single line
readlines()- will return all lines in str element in list
writelines()- from list to file
write()-single line
tell()- current cursor position
seek()- change position of the cursor
'''
```

```python
with open('cpu.log') as config:
    lines = config.readlines()
    data = []
    for line in lines:
        data.append(line.strip())
    print(data)
```
